# Mission 4：算術運算與成績公式

```python
# 一位假想學生的資料
study_hours = 2.5      # 讀書時數
sleep_hours = 7.0      # 睡眠時數
screen_time_hours = 4.0 # 螢幕使用時數
attendance_rate = 90   # 出席率（百分比）

# 分別計算各項影響（方便觀察每個因素的貢獻）
base_score = 50
study_effect = 3 * study_hours
sleep_effect = 2 * sleep_hours
screen_effect = -1.2 * screen_time_hours
attendance_effect = 0.15 * attendance_rate

# 合計總分
quiz_score = base_score + study_effect + sleep_effect + screen_effect + attendance_effect

print("基礎分數：", base_score)
print("讀書貢獻：", study_effect)
print("睡眠貢獻：", sleep_effect)
print("螢幕影響：", screen_effect)
print("出席貢獻：", attendance_effect)
print("--------------------")
print("模擬小考成績：", quiz_score)
```

# Mission 5：條件判斷與自訂函式

```python
# 定義函式：輸入分數，輸出文字描述
def describe_score(score):
    # 判斷分數區間
    if score >= 80:
        result = "表現良好"
    elif score >= 60:
        result = "達到基本標準"
    else:
        result = "需要加強"

    # 回傳結果
    return result

# 呼叫函式並印出結果
print("分數 85：", describe_score(85))
print("分數 70：", describe_score(70))
print("分數 50：", describe_score(50))
```
